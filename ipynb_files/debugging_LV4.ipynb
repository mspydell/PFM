```python
import shutil
import netCDF4 as nc
import numpy as np

fn_IC = '/scratch/PFM_Simulations/restart_roms_data_new/LV4_ocean_rst_202607210000_202607260000.nc'
fn_IC_fixed = '/scratch/PFM_Simulations/restart_roms_data_new/LV4_ocean_rst_202607210000_202607260000_fixed.nc'

# 1. Duplicate the original file
shutil.copyfile(fn_IC, fn_IC_fixed)

# 2. Modify the new file in-place
with nc.Dataset(fn_IC_fixed, 'r+') as ds:
    time_var = ds.variables['ocean_time']
    ts = time_var[:]
    
    # Round to the nearest hour (3600 seconds)
    time_var[:] = np.round(ts / 3600.0) * 3600.0

print(f"Saved modified file to: {fn_IC_fixed}")
```

```python
from datetime import datetime, timedelta

with nc.Dataset(fn_IC_fixed) as ds:
    ts = ds.variables['ocean_time'][:]

t0 = datetime(1999, 1, 1)
time = [t0 + timedelta(seconds=float(t)) for t in ts]

print(time)
```
